# Problem Framing

## Problem
The goal of this project is to predict whether an airline passenger is satisfied or neutral/dissatisfied based on their demographic information, travel details, and airline service-related features.

## Prediction Target
The target variable is **satisfaction**.

The target has two classes:
- **satisfied**
- **neutral or dissatisfied**

## Task Type
This is a **Binary Classification** problem because there are two possible output classes.

## Success Metric
The primary success metric is **F1-score**.

Additional evaluation metrics will include:
- Accuracy
- Precision
- Recall
- Confusion Matrix

```python
import pandas as pd

df = pd.read_csv('train.csv')

df = df.sample(n=10000, random_state=42)

print(df.shape)
df.head()
df.info()

df['satisfaction'].value_counts(normalize=True) * 100

df.isnull().sum()
df.duplicated().sum()

df = df.drop(columns=['id', 'Unnamed: 0'])

df.head()

x = df.drop(columns=['satisfaction'])
y = df['satisfaction']
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x='satisfaction')

plt.title('Passenger Satisfaction Distribution')
plt.xlabel('Satisfaction')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(data=df, x='Class', hue='satisfaction')

plt.title('Passenger Satisfaction by Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.boxplot(data=df, x='satisfaction', y='Flight Distance')

plt.title('Flight Distance by Satisfaction')
plt.xlabel('Satisfaction')
plt.ylabel('Flight Distance')

plt.show()
```
